# Chat Intent Model Training

This notebook trains a local text classification model for the AI Chatbot's intent parser.

```python
import joblib
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neural_network import MLPClassifier

# Synthetic training data
train_data = [
    # add_transaction
    ("spent 500 on food", "add_transaction"),
    ("got my salary 50000", "add_transaction"),
    ("paid electricity bill 1200", "add_transaction"),
    ("bought movie tickets 400", "add_transaction"),
    ("just spent rs 200 on blinkit", "add_transaction"),
    ("transferred 5000 to landlord for rent", "add_transaction"),
    
    # add_goal
    ("save 5000 in 6 months", "add_goal"),
    ("goal for buying a car 500000 12 months", "add_goal"),
    ("i want to save 20000 for emergency in 3 months", "add_goal"),
    ("create a goal to save 10000", "add_goal"),
    
    # investment
    ("where should i invest", "investment"),
    ("recommend mutual funds", "investment"),
    ("how does sip work", "investment"),
    ("investing in stocks vs fd", "investment"),
    ("can you suggest a high risk portfolio", "investment"),
    
    # budget_analysis
    ("am i overspending", "budget_analysis"),
    ("how is my budget looking", "budget_analysis"),
    ("analyze my expenses", "budget_analysis"),
    ("tell me where my money is going", "budget_analysis"),
    
    # tax_advice
    ("how to save tax under 80c", "tax_advice"),
    ("tax deductions for health insurance", "tax_advice"),
    ("itr filing guide", "tax_advice"),
    ("what is the tax on mutual funds", "tax_advice"),
    
    # analysis
    ("what is inflation", "analysis"),
    ("hi bot", "analysis"),
    ("explain compound interest", "analysis"),
    ("who are you", "analysis"),
    ("what is nifty 50", "analysis")
]

docs, labels = [], []
for text, intent in train_data:
    # Duplicate for robust small dataset training
    for _ in range(5):
        docs.append(text)
        labels.append(intent)

print(f"Training on {len(docs)} samples...")
```

```python
# Train Vectorizer and Classifier
vec = TfidfVectorizer(ngram_range=(1, 2), stop_words='english')
X = vec.fit_transform(docs)

clf = MLPClassifier(hidden_layer_sizes=(32, ), max_iter=200, random_state=42)
clf.fit(X, labels)

print("Training Complete. Accuracy on training set:", clf.score(X, labels))
```

```python
# Export Models
os.makedirs('model', exist_ok=True)
joblib.dump(vec, 'model/intent_vec.pkl')
joblib.dump(clf, 'model/intent_clf.pkl')

print("Models saved to model/intent_vec.pkl and model/intent_clf.pkl")
```
